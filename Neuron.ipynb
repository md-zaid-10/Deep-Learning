```python
import torch
import torch.nn as nn
```

```python
'''
0D - scaler (costant value)
1D - vector
2D - Matrix
3D - tensors (multi-dimensional array)
'''
```

```python
# Input

torch.manual_seed(42)
inputs = torch.tensor([1.0, 2.0, 3.0]) # tensors are multi dimensional arrays
```

```python
neuron = nn.Linear(in_features=3, out_features=1)
```

```python
output = neuron(inputs)
```

```python
output
```

```python
## Verify with formula

neuron.weight@inputs + neuron.bias
```
